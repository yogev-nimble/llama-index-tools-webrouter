# WebRouter Tool

This notebook shows `WebRouterToolSpec` on its own and inside an agent. Search and extract need no API key; the agent needs an LLM key (here `OPENAI_API_KEY`).

```python
%pip install llama-index-tools-webrouter llama-index-llms-openai
```

```python
from llama_index.tools.webrouter import WebRouterToolSpec

webrouter_tool = WebRouterToolSpec()
```

## Search

```python
for doc in webrouter_tool.search("LlamaIndex function agents", max_results=3):
    print(doc.metadata["title"], doc.metadata["url"])
```

## Extract

```python
pages = webrouter_tool.extract(["https://docs.python.org/3/whatsnew/index.html"])
print(pages[0].text[:500])
```

## Use the tools in an agent

```python
from llama_index.core.agent.workflow import FunctionAgent
from llama_index.llms.openai import OpenAI

agent = FunctionAgent(
    tools=webrouter_tool.to_tool_list(),
    llm=OpenAI(model="gpt-4o"),
)

response = await agent.run(
    "What changed in the latest Python release? Read the release notes."
)
print(response)
```
